```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f'{self.name}さん：{self.age}歳のあなたは大人です')
    else:
      print(f'{self.name}さん：{self.age}歳のあなたは大人ではありません')

person1 = Human('Yamada', 29)
person2 = Human('Takahashi', 19)

members = [person1, person2]

for member in members:
  member.check_adult()
```
