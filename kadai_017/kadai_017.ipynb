```python
class Human:
  def __init__(self, age):
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f'{self.age}歳のあなたは大人です')
    else:
      print(f'{self.age}歳のあなたは大人ではありません')

yamada = Human(29)
takahashi = Human(19)

members = [yamada, takahashi]

for member in members:
  member.check_adult()
```
